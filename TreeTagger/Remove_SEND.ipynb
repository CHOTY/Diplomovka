```python
import pandas as pd

# 1. Načítanie pôvodného Excel súboru
df = pd.read_excel("outputTreeTagger.xlsx")

# 2. Vytvorenie masky (riadky, kde je Token bodka ".")
mask = (df["Token"] == ".")

# 3. Upravíme Xpos1 a Xpos2-7 pre tieto riadky
df.loc[mask, "Xpos1"] = "Z"
df.loc[mask, ["Xpos2","Xpos3","Xpos4","Xpos5","Xpos6","Xpos7"]] = "-"

# 4. Uloženie výsledku do nového (alebo prepísanie pôvodného) Excel súboru
df.to_excel("outputTreeTagger_updated.xlsx", index=False)
```
