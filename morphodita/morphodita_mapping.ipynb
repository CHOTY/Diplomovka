```python
import pandas as pd
```

```python
# Ilustračná mapovacia tabuľka (MorphoDita prefix -> Stanza Xpos1)
# Upraviť/rozšíriť podľa skutočných prefixov, ktoré sa v Tag vyskytujú.
morphodita_map = {
    # Slovesá
    "VB": "V",  # VB-S---3P-AA---, VB-P---1P-AA---
    "Vp": "V",  # VpXP---XR-AA---
    "Vf": "V",  # Vf--------A----
    "Vc": "V",  # Vc... (ak by bolo)
    "Vs": "V",  # VsIP---XX-AP--- (trpné príčastie)
    
    # Podstatné mená
    "NN": "S",  # NNIS4, NNFS6, NNMP4, NNIP2...

    # Prídavné mená
    "AA": "A",  # AAIS1, AAFS6...

    # Zámená / častice
    "P7": "R",  # P7-X4 (zvratné "sa"), P7-X3...
    "PS": "R",  # PSFS6-P1, PSMP3--3..., privlastňovacie atď.
    "PP": "R",  # PP-P4--1, PPMP7--3..., osobné zámená
    "PD": "P",  # PDIS1, PDMS4... (demonštratívne?)
    "PQ": "P",  # PQ--4... (tvar "čo", "ktorý"?)
    "PW": "P",  # PW...
    "PH": "P",  # PH...
    "P4": "P",  # P4FS4 (ktorý?)
    "P8": "R",  # P8FS7 (svoj...)
    "P5": "R",  # P5FS4 (?)
    
    # Príslovky
    "Dg": "D",  # Dg-------1A----
    "Db": "D",  # Db-------------
    "Co": "D",  # Co------------- (častokrát)

    # Číslovky
    "Cr": "N",  # CrIS6 (radové číslo)
    "Cn": "N",  # Cn-S7 (napr. "päť", "desať"?)
    "C=": "0",  # cislo ako 17
    "Cl": "N",  # štyrmi
    "Cv": "N",  # druhýkrát

    # Spojky, partikuly
    "J^": "O",  # a-1, ani, aj...
    "J,": "O",  # že (J,-------------)
    "TT": "T",  # TT------------- (napr. "no", "tak"?)

    # Predložky
    "RR": "E",  # RR--6, RR--4
    "RV": "E",  # RV--2, RV--6 (napr. "vo"?)

    # Interpunkcia
    "Z:": "Z",  # bodka, čiarka, ...
    "Z,": "Z",  # ak by bolo
    "!":  "Z",  # ak by sa vyskytlo

    # Neznáme, citoslovcia...
    "XX": "T",  # XX------------- (napr. "už" v MorphoDita)
    "X@": "X",  # X@------------- (špeciálne tvary)
    "Db": "D",  # ak je "Db" (e.g. "veľmi"?)

    # Netusim
    "PL": "P",
    "PZ": "P",

}

def morphodita_to_xpos1(tag):
    """Vráti len prvý písmenkový UPOS (podľa Stanza logiky) na základe prefixu MorphoDita tagu."""
    if not isinstance(tag, str):
        return "-"
    t = tag.strip()
    if len(t) < 2:
        return "-"
    
    # Vyberieme prvé 2 znaky. Napr. "VB", "NN", "Z:", "XX", "P7", ...
    prefix2 = t[:2]
    
    # Skúsime nájsť v mapovaní
    if prefix2 in morphodita_map:
        return morphodita_map[prefix2]
    
    # Niekedy prefix "J," atď. Môžeš vyskúšať aj 2+ znaky
    # Alebo ak je to "J," -> "O"
    # Skúsime prvé 3 znaky:
    prefix3 = t[:3]
    if prefix3 in morphodita_map:
        return morphodita_map[prefix3]
    
    # Inak daj "-"
    return "-"
```

```python
# ----------------------------------------------------
# 1. Načítame súbor s MorphoDita tagmi
# (obsahuje stĺpce: Token, Lemma, Tag)
# ----------------------------------------------------
df = pd.read_excel(r"C:\Users\marti\Diplomovka\morphodita\words.xlsx")

# ----------------------------------------------------
# 2. Vytvoríme nový stĺpec Xpos1 s mapovaním
# ----------------------------------------------------
df["Xpos1"] = df["Tag"].apply(morphodita_to_xpos1)

# ----------------------------------------------------
# 3. Uložíme výsledok
# ----------------------------------------------------
df.to_excel(r"C:\Users\marti\Diplomovka\morphodita\words_with_Xpos1.xlsx", index=False)
```
